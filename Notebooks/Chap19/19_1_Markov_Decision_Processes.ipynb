Define a step from the Markov process

```python
def markov_process_step(state, transition_probabilities):
  # TODO -- update the state according to the appropriate transition probabilities
  # One way to do this is to use np.random.choice
  # Get the number of possible states
  n_states = transition_probabilities.shape[0]

  # Select the next state.
  # We choose from the indices 0 to n_states-1,
  # using the probabilities corresponding to the current state.
  # Normalize probabilities to ensure they sum to 1, addressing potential floating point inaccuracies
  normalized_probabilities = transition_probabilities[state] / np.sum(transition_probabilities[state])
  new_state = np.random.choice(n_states, p=normalized_probabilities)

  return new_state
```

Define a Markov one step of a reward process.

```python
def markov_reward_process_step(state, transition_probabilities, reward_structure):
    # Update the state using the transition probabilities for the current state
    n_states = transition_probabilities.shape[0]
    # Normalize probabilities to ensure they sum to 1
    normalized_probabilities = transition_probabilities[state] / np.sum(transition_probabilities[state])
    new_state = np.random.choice(n_states, p=normalized_probabilities)

    # Return a reward of +1 if the Penguin lands on the fish or zero otherwise.
    # The reward_structure array contains 1 at fish indices and 0 elsewhere.
    # Use .item() to extract the scalar value from the numpy array.
    reward = reward_structure[new_state].item()

    return new_state, reward
```

Now let's calculate the return -- the sum of discounted future rewards

```python
def calculate_return(rewards, gamma):
  # TODO -- you write this function
  # It should compute one return for the start of the sequence (i.e. G_1)
  # Replace this line
  # The input 'rewards' is a list/array of future rewards: [r_{t+1}, r_{t+2}, ..., r_T]
  return_val = 0.0

  # Loop through the rewards and apply the discount factor gamma raised to the power of the step index
  for i in range(len(rewards)):
      # i=0 corresponds to r_{t+1}, i=1 to r_{t+2}, etc.
      return_val += (gamma**i) * rewards[i]

  return return_val
```
